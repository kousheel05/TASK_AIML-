Markov Decision Process (MDP)

Definition

A Markov Decision Process (MDP) is a mathematical framework used in Reinforcement Learning (RL) to model decision-making problems where an agent interacts with an environment to maximize rewards.

Why do we use MDP?

Reinforcement Learning problems involve many decisions.

For example:

A robot deciding which way to move.

A self-driving car deciding when to brake.

A game AI deciding its next move.

MDP provides a structured way to represent these problems.

Markov Property

The most important concept in MDP is the Markov Property.

It states:

The future depends only on the current state, not on how you arrived there.
Mathematically,
                  P(St+1∣St)=P(St+1∣S1,S2,…,St)


MDP Flow
Current State

       │
       ▼

Choose Action

       │
       ▼

Environment Changes

       │
       ▼

Next State

       │
       ▼

Receive Reward

       │
       ▼
Repeat

```python
# States
states = ["Start", "Middle", "Goal"]

# Rewards
rewards = {
    "Start": 0,
    "Middle": 1,
    "Goal": 10
}

gamma = 0.9

# Value of moving from Middle to Goal
value = rewards["Middle"] + gamma * rewards["Goal"]

print("State Value:", value)
```
