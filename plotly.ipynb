```python
import pandas as pd
import plotly.express as px
import plotly.express as px

data = {
    "Subject": ["Python", "Java", "C++", "SQL"],
    "Students": [45, 35, 25, 40]
}

fig = px.bar(
    data,
    x="Subject",
    y="Students",
    title="Students in Each Programming Course"
)

fig.show()
```

```python
import plotly.express as px

data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "Sales": [100, 120, 90, 140, 160]
}

fig = px.line(
    data,
    x="Day",
    y="Sales",
    title="Daily Sales"
)

fig.show()
```

```python

data = {
    "Department": ["IT", "HR", "Sales", "Finance"],
    "Employees": [40, 20, 25, 15]
}

fig = px.pie(
    data,
    names="Department",
    values="Employees",
    title="Employee Distribution"
)

fig.show()
```

```python
import pandas as pd
from google.colab import files
uploaded = files.upload()
df = pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
import plotly.express as px

fig = px.bar(
    df,
    x="Sex",
    title="Passenger Count by Gender",
    color="Sex"
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x="Age",
    nbins=30,
    title="Age Distribution of Passengers",
    color="Sex"
)

fig.show()
```

```python
fig = px.pie(
    df,
    names="Survived",
    title="Survival Distribution"
)

fig.show()
```

```python
fig = px.violin(
    df,
    x="Sex",
    y="Age",
    color="Sex",
    box=True,
    title="Age Distribution by Gender"
)

fig.show()
```
