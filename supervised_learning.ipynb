Support Vector Machine (SVM)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Create Dataset
df = pd.DataFrame({
    'Age': [22, 25, 47, 52, 46, 56, 27, 30],
    'Income': [25000, 30000, 80000, 90000, 70000, 100000, 35000, 40000],
    'Buy': [0, 0, 1, 1, 1, 1, 0, 0]
})

X = df[['Age', 'Income']]
y = df['Buy']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = SVC(kernel='linear')
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
df['Prediction'] = model.predict(X)
print("Predictions:", y_pred)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
fig = px.scatter(
    df,
    x='Age',
    y='Income',
    color='Prediction',
    symbol='Prediction',
    title='SVM Prediction Visualization'
)

fig.show()
```

K-Nearest Neighbors (KNN)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Create Dataset
df = pd.DataFrame({
    'Age': [22, 25, 47, 52, 46, 56, 27, 30],
    'Income': [25000, 30000, 80000, 90000, 70000, 100000, 35000, 40000],
    'Buy': [0, 0, 1, 1, 1, 1, 0, 0]
})

X = df[['Age', 'Income']]
y = df['Buy']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Predictions:", y_pred)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

KNN data visualization

```python
import plotly.express as px

fig = px.box(
    df,
    x='Buy',
    y='Income',
    color='Buy',
    points='all',
    title='KNN - Income Distribution'
)

fig.show()
```

Random Forest

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Create Dataset
df = pd.DataFrame({
    'Age': [22, 25, 47, 52, 46, 56, 27, 30],
    'Income': [25000, 30000, 80000, 90000, 70000, 100000, 35000, 40000],
    'Buy': [0, 0, 1, 1, 1, 1, 0, 0]
})

X = df[['Age', 'Income']]
y = df['Buy']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Predictions:", y_pred)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

Random Forest visualization

```python
import pandas as pd
import plotly.express as px

df = pd.DataFrame({
    'Age':[22,25,47,52,46,56,27,30],
    'Income':[25000,30000,80000,90000,70000,100000,35000,40000],
    'Buy':[0,0,1,1,1,1,0,0]
})

fig = px.histogram(
    df,
    x='Income',
    color='Buy',
    barmode='group',
    title='Random Forest Dataset'
)

fig.show()
```

Polynomial Regression

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Create Dataset
poly_df = pd.DataFrame({
    'X': [1, 2, 3, 4, 5],
    'Y': [1, 4, 9, 16, 25]
})

# Features and Target
X = poly_df[['X']]
y = poly_df['Y']

# Polynomial Features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Train Model
model = LinearRegression()
model.fit(X_poly, y)

# Predict
new = poly.transform([[6]])
prediction = model.predict(new)

print("Prediction for X=6:", prediction[0])
```

Polynomial data visualization

```python
fig = px.scatter(
    poly_df,
    x='X',
    y='Y',
    title='Polynomial Regression Dataset',
    color='Y',
    size='Y'
)

fig.show()
```

Decision Tree

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Create Dataset
df = pd.DataFrame({
    'Age': [22, 25, 47, 52, 46, 56, 27, 30],
    'Income': [25000, 30000, 80000, 90000, 70000, 100000, 35000, 40000],
    'Buy': [0, 0, 1, 1, 1, 1, 0, 0]
})

X = df[['Age', 'Income']]
y = df['Buy']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Predictions:", y_pred)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

Decision Tree Visualization

Linear Regression

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Create Dataset
lr_df = pd.DataFrame({
    'Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Marks': [15, 25, 35, 45, 55, 65, 75, 85]
})

X = lr_df[['Hours']]
y = lr_df['Marks']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)

prediction = model.predict([[9]])
lr_df['Predicted_Marks'] = model.predict(X)
print("Prediction for 9 Hours:", prediction[0])
```

```python
fig = px.scatter(
    lr_df,
    x='Hours',
    y='Marks',
    trendline='ols',
    title='Linear Regression',
    color='Predicted_Marks',
    size='Predicted_Marks'
)

fig.show()
```
