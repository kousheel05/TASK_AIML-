Series

```python
import numpy as np
import pandas as pd
```

```python
s = pd.Series([1,3,4,8,9,5])
print(s)
print(s.describe())
print(s.mean())
print(s.std())
print(s.min())
print(s.max())
```

```python
s1 = pd.Series([1,2,3,4,5,8])
map = s1.map (lambda x:x ** 2)
print(map)
```

```python
s11 = pd.Series([1,1,1,2,2,5,4,5,4,5])
app = s1.apply(lambda x:x % 2)
print(app)
sort = s11.sort_values()
print(sort)
```

```python
s2 = pd.Series([2,4,6,np.nan,7,0])
print(s2.isnull())
s22 = pd.Series([44,5,1,np.nan,12,np.nan,24,55,46,])
print(s22.notnull())
print(s2.iloc[5])
print(s22.iloc[-4])
print(s2.aggregate(['sum','std','mean']))
```
