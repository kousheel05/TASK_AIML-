```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

uploaded = files.upload()
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name, encoding='latin1')
df.info()
```

```python
df.head(10)
```

```python
HS = df.sort_values("subscribers", ascending=False).head(10)
plt.figure(figsize=(10,5))
plt.bar(HS["Youtuber"], HS["subscribers"])
plt.xlabel("Youtuber")
plt.ylabel("Subscribers")
plt.xticks(rotation=90)
plt.title("Top 10 YouTube Channels by Subscribers")
plt.show()
```

```python
hs = df.sort_values("video views", ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(x="Youtuber", y="video views", data=hs)
plt.xlabel("YouTube Channels")
plt.ylabel("Video Views")
plt.title("Top 10 YouTube Channels by Video Views")
plt.xticks(rotation=90)
plt.show()
```

```python
top_10_countries = df["Country"].value_counts().head(10).reset_index()
top_10_countries.columns = ["Country", "Count"]
plt.figure(figsize=(10,5))
sns.barplot(data=top_10_countries, x="Country", y="Count")
plt.xlabel("Countries")
plt.ylabel("Number of YouTube Creators")
plt.title("Top 10 Countries with Highest Number of YouTube Creators")
plt.xticks(rotation=90)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

uploaded = files.upload()
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name, encoding='latin1')
df.info()

HS = df.sort_values("subscribers", ascending=False).head(10)
plt.figure(figsize=(10,5))
plt.bar(HS["Youtuber"], HS["subscribers"])
plt.xlabel("Youtuber")
plt.ylabel("Subscribers")
plt.xticks(rotation=90)
plt.title("Top 10 YouTube Channels by Subscribers")
plt.show()

highest_views = df.sort_values("video views", ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(x="Youtuber", y="video views", data=highest_views)
plt.xlabel("YouTube Channels")
plt.ylabel("Video Views")
plt.title("Top 10 YouTube Channels by Video Views")
plt.xticks(rotation=90)
plt.show()

top_10_countries = df["Country"].value_counts().head(10).reset_index()
top_10_countries.columns = ["Country", "Count"]
plt.figure(figsize=(10,5))
sns.barplot(data=top_10_countries, x="Country", y="Count")
plt.xlabel("Countries")
plt.ylabel("Number of YouTube Creators")
plt.title("Top 10 Countries with Highest Number of YouTube Creators")
plt.xticks(rotation=90)
plt.show()
```
