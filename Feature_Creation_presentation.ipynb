Feature Creation

Feature Creation is the process of creating new features (columns) from existing data to help Machine Learning models learn better patterns.

Simple Definition:-

"We create new useful information from the existing data."

Why is Feature Creation Important?

✅ Improves model accuracy

✅ Helps identify hidden patterns

✅ Makes data more meaningful

✅ Enhances prediction performance

| Type                        | Example                  |
| --------------------------- | ------------------------ |
| Mathematical Transformation | Total = Price × Quantity |
| Combining Features          | First Name + Last Name   |
| Extracting Information      | Email → Domain           |
| Domain-Based Features       | BMI = Weight ÷ Height²   |


Example
| Student | Math | Science |
| ------- | ---- | ------- |
| John    | 80   | 90      |
| Jane    | 75   | 85      |
| Sam     | 95   | 88      |



Applying Feature creation techniques :- New Feature Created

| Student | Math | Science | **Total Marks** | **Average Marks** |
| ------- | ---- | ------- | --------------- | ----------------- |
| John    | 80   | 90      | **170**         | **85**            |
| Jane    | 75   | 85      | **160**         | **80**            |
| Sam     | 95   | 88      | **183**         | **91.5**          |


```python
import pandas as pd

data = {
    "First_Name": ["John", "Jane", "Sam"],
    "Last_Name": ["Doe", "Smith", "Brown"],
    "Math": [80, 75, 95],
    "Science": [90, 85, 88],
    "Email": ["john@gmail.com", "jane@yahoo.com", "sam@outlook.com"],
    "Weight": [60, 70, 80],
    "Height": [1.6, 1.7, 1.8]
}

df = pd.DataFrame(data)
print(df)
```

```python
# 1. Mathematical Transformation
df["Total_Marks"] = df["Math"] + df["Science"]
print(df)
```

```python
# 2. Combining Features
df["Full_Name"] = df["First_Name"] + " " + df["Last_Name"]
print(df)
```

```python
# 3. Extracting Information
df["Email_Domain"] = df["Email"].str.split("@").str[1]
print(df)
```

```python
# 4. Domain-Based Feature
df["BMI"] = df["Weight"] / (df["Height"] ** 2)
print(df)
```

```python
import plotly.express as px

fig = px.bar(
    df,
    x="Full_Name",
    y="Total_Marks",
    title="Total Marks of Students",
    text="Total_Marks",
    color="Total_Marks",
    color_continuous_scale="Blues"
)

fig.update_traces(textposition="outside")

fig.update_layout(
    xaxis_title="Students",
    yaxis_title="Total Marks",
    template="plotly_white"
)

fig.show()
```
