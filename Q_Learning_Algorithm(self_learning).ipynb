Q-Learning Algorithm

Definition

Q-Learning is a model-free Reinforcement Learning algorithm that learns the best action to take in each state by calculating Q-values (Quality Values).

Simple Definition: Q-Learning is an algorithm that helps an agent learn the best action in every state by maximizing future rewards, without knowing how the environment works in advance.

Why is it called Q-Learning?

The letter Q stands for Quality.

The Q-value measures the quality (goodness) of taking an action in a particular state.

Q(State,Action)

It answers the question:

"If I take this action in this state, how good will it be in terms of future rewards?"

What is a Q-Table?

A Q-Table stores the Q-value for every State–Action pair.

Example:

| State | Left | Right | Up | Down |
| ----- | ---- | ----- | -- | ---- |
| A     | 5    | 8     | 4  | 6    |
| B     | 2    | 10    | 7  | 5    |
| Goal  | 0    | 0     | 0  | 0    |



The agent always chooses the action with the highest Q-value.

```python
import numpy as np

# Q-table: 3 states and 2 actions
Q = np.zeros((3, 2))

learning_rate = 0.8
gamma = 0.9

# Current state = 0, action = 1
state = 0
action = 1

# Reward received
reward = 10

# Next state
next_state = 1

# Update Q-value
Q[state, action] = Q[state, action] + learning_rate * (
    reward + gamma * np.max(Q[next_state]) - Q[state, action]
)

print("Updated Q-table:")
print(Q)
```
