Write a Python program using NumPy to perform the following:

Marks = 65, 78, 90, 55, 88

Tasks:

Create a NumPy array using these marks

Find the average marks

Find the highest marks

Find the lowest marks

```python
p = np.array([65,78,90,55,88])
avg = np.average(p)
hih = np.max(p)
low = np.min(p)
print(avg)
print(hih)
print(low)
```

```python
import numpy as np
p = np.array(list(map(int, input("Enter marks separated by spaces: ").split())))
avg = np.average(p)
hih = np.max(p)
low = np.min(p)
sum = np.sum(p)
sh = np.shape(p)
print("Average:", avg)
print("Highest:", hih)
print("Lowest:", low)
print("sum:",sum)
print("shape:",sh)
```
