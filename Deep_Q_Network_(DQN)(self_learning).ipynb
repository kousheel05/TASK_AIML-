Deep Q Network (DQN)

Definition

A Deep Q Network (DQN) is a Deep Reinforcement Learning algorithm that combines Q-Learning with a Deep Neural Network (DNN).

Instead of storing Q-values in a Q-table, DQN uses a neural network to estimate the Q-values for each action.

Simple Definition: DQN is an improved version of Q-Learning that uses a neural network instead of a Q-table, allowing it to solve problems with very large or complex state spaces.

Why Do We Need DQN?

Q-Learning works well for small environments.

Example:

A robot moving in a 3 × 3 grid.

The Q-table is small:
| State | Left | Right | Up | Down |
| ----- | ---- | ----- | -- | ---- |
| S1    | 0    | 5     | 2  | 4    |
| S2    | 3    | 8     | 1  | 6    |


Q-Learning vs DQN
| Q-Learning                         | DQN                                                       |
| ---------------------------------- | --------------------------------------------------------- |
| Uses a Q-table                     | Uses a Deep Neural Network                                |
| Good for small state spaces        | Good for very large state spaces                          |
| Memory grows with number of states | Memory depends mainly on the network size                 |
| Cannot process images directly     | Can learn directly from images or high-dimensional inputs |


Architecture of DQN

          Current State
                │
                ▼
      Deep Neural Network
                │
                ▼
      Predicted Q-values
                │
                ▼
      Choose Best Action
                │
                ▼
      Environment
                │
                ▼
     Reward + Next State
                │
                ▼
      Update Neural Network

DQN Algorithm

Initialize the neural network.

Initialize the target network.

Initialize replay memory.

Observe the current state.

Choose an action using an ε-greedy policy.

Perform the action.

Receive reward and next state.

Store the experience in replay memory.

Randomly sample a batch of experiences.

Train the neural network.

Periodically update the target network.

Repeat.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Build the DQN model
model = Sequential([
    Dense(24, activation="relu", input_shape=(4,)),
    Dense(24, activation="relu"),
    Dense(2, activation="linear")  # Q-values for 2 actions
])

# Compile the model
model.compile(optimizer="adam", loss="mse")

# Example input state
state = tf.constant([[0.2, 0.5, 0.1, 0.9]])

# Predict Q-values
q_values = model.predict(state)

print("Predicted Q-values:")
print(q_values)
```
