```python
dms = sns.load_dataset("diamonds")
dms.info()
dms.head()
sns.barplot(x="cut", y="price", data=dms)
sns.barplot(x="cut", y="price", hue="color", data=dms)
plt.show()
```

```python
hexp = sns.load_dataset("healthexp")
hexp.info()
hexp.head()
print(hexp.columns)
sns.scatterplot(data=hexp,x="Year", y="Spending_USD", hue="Country")
plt.title("year and money")
plt.show()
```

```python
sns.barplot(data=hexp, x="Country", y="Spending_USD")
plt.show()
```

```python
sns.boxplot(data=hexp, x="Country", y="Spending_USD")
plt.show()
```

```python
sns.histplot(data=hexp, x="Spending_USD")
plt.show()
```

```python
corr = hexp[["Year", "Spending_USD", "Life_Expectancy"]].corr()
sns.heatmap(corr, annot=True)
plt.show()
```

```python
sns.violinplot(data=hexp, x="Country", y="Spending_USD")
plt.show()
```
