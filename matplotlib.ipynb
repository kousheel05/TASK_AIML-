```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D"]
marks = [80, 95, 70, 85]

data = sorted(zip(students, marks), key=lambda x: x[1], reverse=True)

students, marks = zip(*data)

plt.bar(students, marks)
plt.title("Marks in Descending Order")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import shutil
from google.colab import files

uploaded = files.upload()
df = pd.read_csv("myhospitals-contact-details (1).csv", encoding="latin1")
os.makedirs("Graphs", exist_ok=True)

state_counts = df['State'].value_counts()
plt.figure(figsize=(8,5))
plt.bar(state_counts.index, state_counts.values)
plt.title("Hospitals by State")
plt.xlabel("State")
plt.ylabel("Number of Hospitals")
plt.savefig("Graphs/01_Hospitals_by_State.png", dpi=300, bbox_inches='tight')
plt.show()

state_counts_desc = df['State'].value_counts().sort_values(ascending=False)
plt.figure(figsize=(8,5))
plt.bar(state_counts_desc.index, state_counts_desc.values)
plt.title("Hospitals by State (Descending)")
plt.xlabel("State")
plt.ylabel("Number of Hospitals")
plt.savefig("Graphs/02_Hospitals_by_State_Descending.png", dpi=300, bbox_inches='tight')
plt.show()

sector_counts = df['Sector'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(
    sector_counts.values,
    labels=sector_counts.index,
    autopct='%1.1f%%'
)
plt.title("Hospital Sector Distribution")
plt.savefig("Graphs/03_Sector_Distribution.png", dpi=300, bbox_inches='tight')
plt.show()
plt.figure(figsize=(8,5))
plt.hist(df['Beds'].dropna(), bins=20)
plt.title("Beds Distribution")
plt.xlabel("Beds")
plt.ylabel("Frequency")
plt.savefig("Graphs/04_Beds_Distribution.png", dpi=300, bbox_inches='tight')
plt.show()

plt.figure(figsize=(8,6))
plt.scatter(df['Longitude'], df['Latitude'])
plt.title("Hospital Locations")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.savefig("Graphs/05_Hospital_Locations.png", dpi=300, bbox_inches='tight')
plt.show()


plt.figure(figsize=(8,5))
plt.barh(state_counts.index, state_counts.values)
plt.title("Hospitals by State (Horizontal)")
plt.savefig("Graphs/06_Hospitals_Horizontal_Bar.png", dpi=300, bbox_inches='tight')
plt.show()


top10 = df.sort_values(by='Beds', ascending=False).head(10)

plt.figure(figsize=(12,6))
plt.bar(top10['Hospital name'], top10['Beds'])
plt.xticks(rotation=90)
plt.title("Top 10 Hospitals by Beds")
plt.savefig("Graphs/07_Top10_Hospitals_by_Beds.png", dpi=300, bbox_inches='tight')
plt.show()
print(df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
shutil.make_archive("Hospital_Graphs", "zip", "Graphs")
files.download("Hospital_Graphs.zip")
```
