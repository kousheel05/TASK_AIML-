```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Patient_ID" : [101, 102, 103, 104, 105, 106, 107],
    "Name" : ['sruthi','hema','ashok','rohith','venkatesh','mohith','Azmath'],
    "Age" : [23,32,21,15,65,51,22],
    "Disease" : ['Cardiovascular Disease', 'Diabetes','Asthma','Cancer','COVID-19','Alzheimers','Anxiety Disorders'],
    "Treatment_Cost" : [100000, 50000, 10000, 5000, 200000, 80000, 25000],
    "Days_Admitted" : [20, 3, 5, 2, 30, 60, 7]

}
df = pd.DataFrame(data)
df
```

```python
print(df.info())
```

```python
print(df.isnull())
print(df.fillna(0))
```

```python
print(df.dropna())
```

```python
cost=df['Treatment_Cost'] / df['Days_Admitted']
print(cost)
```

```python
avg_cost = df["Treatment_Cost"].mean()
above_avg = df[df["Treatment_Cost"]>avg_cost]
print(avg_cost)
print(above_avg)
```

```python
sorted_df = df.sort_values(by='Treatment_Cost', ascending=False)
print(sorted_df)
```

```python
disease_avg = df.groupby('Disease')["Treatment_Cost"].mean()
print(disease_avg)
```

```python
costs = df["Treatment_Cost"].values
print(costs)
```

```python
import numpy as np

costs = df["Treatment_Cost"].values

print("Mean:", np.mean(costs))
print("Median:", np.median(costs))
print("Minimum:", np.min(costs))
print("Maximum:", np.max(costs))
print("Sum:", np.sum(costs))
print("Standard Deviation:", np.std(costs))
print("Variance:", np.var(costs))
```

```python
most_expensive = df.loc[df["Treatment_Cost"].idxmax()]
print(most_expensive)
```

```python
print("\n========== HOSPITAL REPORT ==========\n")

print("Total Patients:", len(df))
print("Average Age:", round(df["Age"].mean(), 2))
print("Average Treatment Cost:", round(df["Treatment_Cost"].mean(), 2))
print("Average Days Admitted:", round(df["Days_Admitted"].mean(), 2))

print("\n----- Treatment Cost Analysis -----")
print("Total Treatment Cost:", df["Treatment_Cost"].sum())
print("Highest Treatment Cost:", df["Treatment_Cost"].max())
print("Lowest Treatment Cost:", df["Treatment_Cost"].min())

print("\n----- Most Expensive Treatment -----")
most_expensive = df.loc[df["Treatment_Cost"].idxmax()]
print(most_expensive)

print("\n----- Patients with Above Average Cost -----")
avg_cost = df["Treatment_Cost"].mean()
print(df[df["Treatment_Cost"] > avg_cost])

print("\n----- Disease-wise Average Treatment Cost -----")
print(df.groupby("Disease")["Treatment_Cost"].mean())

print("\n========== END OF REPORT ==========")
```
