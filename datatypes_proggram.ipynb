list program

```python
stud = ['kousheel','ajith','deepka','bilal']
print(stud)
stud.append('harsha')
print(stud)
stud.insert(3 ,'srinu')
print(stud)
stud.pop(1)
print(stud)
stud.sort()
stud.reverse()
print(stud)
```

program for tuple

```python
a = ("apple", "banana", "mango")
print(a)
print(len(a))
print((a).index('apple'))
print((a).count('apple'))
```

program for sets

```python
num1 = {1,2,3,4}
num2 = {3,4,5,6}
print(num1)
print(num2)
union = num1.union(num2)
print(union)
intersection = num1.intersection(num2)
print(intersection)
dif = num1.difference(num2)
print(dif)
```

dictionary functions

```python
student = {
    "name": "Kousheel",
    "age": 21,
    "branch": "Cybersecurity"
}

print(student)
print(student.keys())
print(student.values())
print(student.items())
```
