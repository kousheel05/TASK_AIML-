Reinforcement Learning (RL)

Definition

Reinforcement Learning (RL) is a type of Machine Learning in which an agent learns by interacting with an environment. The agent performs actions, receives rewards or penalties, and gradually learns the best actions to maximize the total reward.

Simple Definition: Reinforcement Learning is a trial-and-error learning process where an agent learns the best decision by receiving rewards and punishments.

Why is Reinforcement Learning Used?

Traditional machine learning learns from labeled data, but in many real-world problems:

There are no correct answers (labels).

The agent must learn by experience.

The goal is to maximize long-term rewards.


RL is used when an agent must make a sequence of decisions.

Components of Reinforcement Learning

RL has five main components.

1. Agent

The Agent is the learner or decision-maker.

Examples:

Robot

Self-driving car

Game-playing AI

Trading bot

2. Environment

The Environment is everything the agent interacts with.

Examples:

Chess board

Road for a self-driving car

Video game

Warehouse

3. State (S)

A State is the current situation of the agent.

Example:

A robot is standing at position (2,3) in a grid.

Current Position = State

4. Action (A)

An Action is what the agent chooses to do.

Examples:

Move Left

Move Right

Pick an object

Accelerate

Brake

5. Reward (R)

A Reward is the feedback received after taking an action.

Examples:

Reach goal → +100

Hit wall → -20

Every move → -1

The agent tries to maximize the total reward.

Reinforcement Learning Workflow

Environment

      │
      ▼

Current State

      │
      ▼

Agent Chooses Action

      │
      ▼

Environment Changes

      │
      ▼

Reward Received

      │
      ▼

Agent Learns

      │
      └──────────────► Repeat

Types of Reinforcement Learning

The agent receives a reward for good actions.

Example:

Dog sits when commanded.

Reward:

Give a biscuit.

The dog learns to sit.

2. Negative Reinforcement

The agent removes an unpleasant condition by taking the correct action.

Example:

A robot moves away from an obstacle to stop an alarm.

The robot learns to avoid obstacles.

```python
import random

actions = ["Left", "Right", "Up", "Down"]

for episode in range(5):
    action = random.choice(actions)

    if action == "Right":
        reward = 10
    elif action == "Up":
        reward = 5
    else:
        reward = -2

    print(f"Episode {episode+1}")
    print("Action:", action)
    print("Reward:", reward)
    print()
```
